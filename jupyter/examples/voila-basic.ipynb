# Hello, Voilà 👋

This page is an ordinary notebook with the **code hidden**. Use the controls below: every change runs Python live, in your own kernel.

*Open it as an app:* in JupyterLab click the **Voilà** button in the notebook toolbar, or use the address
`/jupyter/user/<your-name>/voila/render/examples/voila-basic.ipynb`. Open the notebook itself to read the code behind each section.

```python
# Everything a Voilà page needs is already installed: ipywidgets for the controls, matplotlib for charts.
import ipywidgets as w
import numpy as np
import matplotlib.pyplot as plt
```

## 1. A greeting
Type a name and tick the box.

```python
# `interact` builds the controls from the function arguments and re-runs the function whenever one changes.
# Whatever the function prints or displays appears below the controls.
@w.interact(name=w.Text(value="world", description="Your name"),
            shout=w.Checkbox(value=False, description="Shout"))
def greet(name, shout):
    message = f"Hello, {name or 'stranger'}!"
    print(message.upper() if shout else message)
```

## 2. A live chart
Drag the sliders; the chart is redrawn each time you let go.

```python
# continuous_update=False redraws once when you release the slider instead of on every pixel (smoother for heavier charts).
@w.interact(frequency=w.FloatSlider(value=2, min=0.5, max=8, step=0.5, description="Frequency", continuous_update=False),
            amplitude=w.FloatSlider(value=1, min=0.2, max=2, step=0.1, description="Amplitude", continuous_update=False))
def wave(frequency, amplitude):
    x = np.linspace(0, 2 * np.pi, 400)
    fig, ax = plt.subplots(figsize=(7, 3))
    ax.plot(x, amplitude * np.sin(frequency * x), color="#2a7de1", linewidth=2)
    ax.set_ylim(-2.2, 2.2)
    ax.grid(alpha=0.3)
    ax.set_title(f"y = {amplitude} · sin({frequency} · x)")
    plt.show()
```

## How it works

- Voilà runs the notebook **top to bottom in a fresh kernel for each visitor**, then shows only the outputs.
- The controls are *widgets*: the browser tells the kernel, the kernel runs your Python, and the result comes straight back.
